```python
# %% Cellule 1 : Configuration et Importations
import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Inclusion du dossier 'src' dans le chemin système
sys.path.append('../')
from src.models import EcoEbolaClustering

# %% Cellule 2 : Chargement des données préparées
df_zone = pd.read_csv('../data/donnees_zones_preparees.csv', sep=';')

# %% Cellule 3 : Recherche du nombre optimal de clusters (Méthode du coude)
from sklearn.cluster import KMeans
model_explorer = EcoEbolaClustering()
scaled_data = model_explorer.fit_prepare_data(df_zone)

wcss = []
for i in range(1, 7):
    km = KMeans(n_clusters=i, random_state=42, n_init=10)
    km.fit(scaled_data)
    wcss.append(km.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(range(1, 7), wcss, marker='o', linestyle='--', color='#2c3e50')
plt.title('Méthode du Coude (Optimisation des ressources)')
plt.xlabel('Nombre de Clusters')
plt.ylabel('WCSS (Inertie Intra-classe)')
plt.grid(True)
plt.show()

# %% Cellule 4 : Application de K-Means et Profilage
# On initialise le pipeline avec 3 profils types de vulnérabilité
pipeline = EcoEbolaClustering(n_clusters=3)
df_results = pipeline.apply_kmeans(df_zone)

# Extraction des caractéristiques de chaque groupe
profils = pipeline.get_cluster_profiles(df_results, 'cluster_kmeans')
print("\n--- Profils Moyens Éco-Épidémiologiques des Clusters ---")
print(profils)

# %% Cellule 5 : Visualisation Spatiale Conceptuelle
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_results, 
    x='longitude', 
    y='latitude', 
    hue='cluster_kmeans', 
    palette='Set1', 
    size='total_cases', 
    sizes=(40, 400),
    alpha=0.8
)
plt.title('Segmentation Spatiale des Zones de Santé (Ituri / Nord-Kivu)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend(title='Profil de Risque (Cluster)')
plt.grid(True, linestyle=':')
plt.show()
```
