```python
# %% Cellule 1 : Importations
import pandas as pd
import numpy as np

# %% Cellule 2 : Chargement des données enrichies
print("Chargement de la base de données éco-épidémiologique...")
df = pd.read_csv('../data/donnees_ebola_enrichies.csv', sep=';')
df.head()

# %% Cellule 3 : Agrégation par Zone de Santé
# Comme le fichier contient un historique temporel, on agrège par zone pour obtenir la situation globale
agg_rules = {
    'province': 'first',
    'total_cases': 'max',          # Représente le cumul maximal atteint
    'total_deaths': 'max',         # Cumul de décès maximal
    'population_density': 'first', # Invariant environnemental
    'forest_coverage_percent': 'first',
    'water_proximity_index': 'first',
    'latitude': 'first',
    'longitude': 'first'
}

df_zone = df.groupby('health_zone').agg(agg_rules).reset_index()
print(f"Structure de l'analyse : {df_zone.shape[0]} zones de santé identifiées.")

# %% Cellule 4 : Exportation des données prêtes pour le Clustering
df_zone.to_csv('../data/donnees_zones_preparees.csv', index=False, sep=';')
print("Données agrégées et sauvegardées avec succès.")
```
